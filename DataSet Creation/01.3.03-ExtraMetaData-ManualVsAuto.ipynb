```python
import numpy as np
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
fpath = r"d:\DataSets-Local\1405-04-03\Merged_Dataset_2026-06-24T15-15-22\Extra_MetaData\PS_selected.csv"
metadata_manual = pd.read_csv(fpath)
```

```python
fpath = r"D:/DataSets-Local/1405-04-03/Merged_Dataset_2026-06-24T15-15-22/Extra_MetaData/auto_picker.pkl"
metadata = pd.read_pickle(fpath)
```

```python
df = pd.merge(
    left=metadata_manual,
    right=metadata,
    on="trace_name",
)
# msk = df[['Manual_Pick_S','Manual_Pick_P']].notna().all(axis=1)
# df = df[msk]
```

```python
metadata_dict = {
    col_name: np.array(df[col_name].explode().dropna().to_list())
    for col_name in df.columns
    if (
        col_name.endswith('P')
        or
        col_name.endswith('S')
    )
}
```

```python
df_tmp = df[[
    'EQTransformer_scedc_0.1_P',
    'EQTransformer_instance_0.1_P',
    'EQTransformer_original_0.1_P',
    'PhaseNet_original_0.3_P',
]]
bins = np.arange(-5, 5.5, 0.5) - 0.25
for col_name in df_tmp.columns:
        # col_name = 'BasicPhaseAE_scedc_0.3_P'
        a = np.array(df_tmp[col_name].explode().dropna().to_list())
        a -= 500
        a = a / 100
        plt.hist(a, bins=bins, edgecolor='k')
        plt.title(col_name)
        plt.show()
```

```python
for col_name in metadata.columns:
    if (
            col_name.endswith('P')
            # or
            # col_name.endswith('S')
        ): 
        # col_name = 'BasicPhaseAE_scedc_0.3_P'
        a = np.array(metadata[col_name].explode().dropna().to_list())
        a -= 500
        a = a / 100
        # bins = np.arange(0, 6000, 1)
        plt.hist(a, bins=100, edgecolor='k')
        plt.title(col_name)
        plt.show()
```

```python
for phase in [
        "_P",
        # "_S",
    ]:
    d = {
        k: (v - 500) / 100
        for k, v in metadata_dict.items()
        if k.endswith(phase)
    }
    x = d.values()
    tick_labels = d.keys()
    print(tick_labels)
    fig, ax = plt.subplots(figsize=(12, 5))

    
    bp = ax.boxplot(
        x = x,
        tick_labels=tick_labels,
        flierprops={'marker': 'o',
                    # 'markerfacecolor': 'red', 
                    # 'markersize': 8,
                    'alpha': 0.1,
        },
        patch_artist=True,
        showmeans=True,
        meanline=True
    )

    for i, data in enumerate(x, start=1):
        n = len(data)
        mean = np.mean(data)
        mode = stats.mode(data)
        ax.text(
            i-0.5,
            0,
            f'n: {n}, mean: {mean:.0f}\nc: {mode.count:<7}, mode: {mode.mode:.0f}',
            rotation='vertical',
            # ha='center', va='bottom', 
            fontsize=10, fontweight='bold'
        )

    plt.ylim(top=40)
    plt.xticks(rotation=45, ha='right')
    plt.show()
```
